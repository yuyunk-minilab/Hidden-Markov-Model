```python
import numpy as np
import math
from hmmlearn import hmm

# ============================================
# 1. Membuat Categorical HMM
# ============================================
model = hmm.CategoricalHMM(
    n_components=2,
    init_params=""
)

# ============================================
# 2. Probabilitas awal state
# ============================================
model.startprob_ = np.array([
    0.6,
    0.4
])

# ============================================
# 3. Probabilitas transisi
# ============================================
model.transmat_ = np.array([
    [0.7, 0.3],
    [0.4, 0.6]
])

# ============================================
# 4. Probabilitas emisi
# ============================================
model.emissionprob_ = np.array([
    [0.1, 0.4, 0.5],
    [0.6, 0.3, 0.1]
])

# ============================================
# 5. Observasi
#    0 = kategori observasi pertama
#    1 = kategori observasi kedua
#    2 = kategori observasi ketiga
# ============================================
X = np.array([
    [0]
])

# ============================================
# 6. Hitung log probability
# ============================================
log_probability = model.score(X)

# ============================================
# 7. Ubah log probability menjadi probability
# ============================================
probability = math.exp(log_probability)

print("Log probability :", log_probability)
print("Probability     :", probability)
```

```python
import numpy as np
import math
from hmmlearn import hmm

# Membuat model Categorical HMM
model = hmm.CategoricalHMM(
    n_components=2,
    init_params=""
)

# Probabilitas awal state
model.startprob_ = np.array([
    0.6,
    0.4
])

# Probabilitas transisi
model.transmat_ = np.array([
    [0.7, 0.3],
    [0.4, 0.6]
])

# Probabilitas emisi
# Kolom: clean, walk, shop
model.emissionprob_ = np.array([
    [0.1, 0.4, 0.5],
    [0.6, 0.3, 0.1]
])

# Mapping observasi
# clean = 0
# walk  = 1
# shop  = 2

observations = np.array([
    [0],  # clean
    [0],  # clean
    [1],  # walk
    [1],  # walk
    [2]   # shop
])

# Menghitung log probability
logprob = model.score(observations)

# Mengubah log probability menjadi probability
probability = math.exp(logprob)

print("Log probability :", logprob)
print("Probability     :", probability)
```
